```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
text = "West Monroe was referenced in Mr. Deeds."
```

```python
doc = nlp(text)
```

```python
for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
ruler = nlp.add_pipe("entity_ruler")
```

```python
nlp.analyze_pipes()
```

```python
patterns = [
    {"label": "GPE","pattern":"West Monroe"}
]
```

```python
ruler.add_patterns(patterns)
```

```python
doc2 = nlp(text)
for ent in doc2.ents:
    print(ent.text,ent.label_)
```
