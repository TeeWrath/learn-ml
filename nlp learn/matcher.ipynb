```python
import spacy
from spacy.matcher import Matcher
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
matcher = Matcher(nlp.vocab)
pattern = [{"LIKE_EMAIL": True}]
matcher.add("EMAIL_ADDRESS",[pattern])
```

```python
doc = nlp("This is my email address bitch : subbu.banjo@rapper.com")
matches = matcher(doc)
```

```python
print(matches)
```

```python
# 0th index means the tuple above, and its 0th index means the long number 1657 something...
print(nlp.vocab[matches[0][0]].text)
```

```python
with open("data/wiki_mlk.txt","r") as f:
    text = f.read()
matcher = Matcher(nlp.vocab)
pattern = [{"POS": "PROPN","OP":"+"}]
matcher.add("PROPER_NOUNS",[pattern])
doc = nlp(text)
matches = matcher(doc)
print(len(matches))
for match in matches[:10]:
    print(match,doc[match[1]:match[2]])
```

```python
with open("data/wiki_mlk.txt","r") as f:
    text = f.read()
matcher = Matcher(nlp.vocab)
pattern = [{"POS": "PROPN","OP":"+"}]
matcher.add("PROPER_NOUNS",[pattern],greedy="LONGEST")
doc = nlp(text)
matches = matcher(doc)
print(len(matches))
for match in matches[:10]:
    print(match,doc[match[1]:match[2]])
```

```python
# To sort all of the above in correc torder
with open("data/wiki_mlk.txt","r") as f:
    text = f.read()
matcher = Matcher(nlp.vocab)
pattern = [{"POS": "PROPN","OP":"+"}]
matcher.add("PROPER_NOUNS",[pattern],greedy="LONGEST")
doc = nlp(text)
matches = matcher(doc)
matches.sort(key= lambda x:x[1])
print(len(matches))
for match in matches[:10]:
    print(match,doc[match[1]:match[2]])
```

```python
# To find token group with next word as a verb
with open("data/wiki_mlk.txt","r") as f:
    text = f.read()
matcher = Matcher(nlp.vocab)
pattern = [{"POS": "PROPN","OP":"+"},{"POS":"VERB"}]
matcher.add("PROPER_NOUNS",[pattern],greedy="LONGEST")
doc = nlp(text)
matches = matcher(doc)
matches.sort(key= lambda x:x[1])
print(len(matches))
for match in matches[:10]:
    print(match,doc[match[1]:match[2]])
```
